- 2.1 자연어 처리란
    - 사람들이 일상에사 사용하는 형식의 말을 자연어라 함, 자연어 처리(NLP)를 문자 그대로 해석하면 자연어를 처리하는 분야, 우리의 말을 컴퓨터에게 이해시키기 위한 기술
    - 자연어 처리가 추구하는 목표는 사람의 말을 컴퓨터가 이해하도록 만들어서 컴퓨터가 우리에게 도움이 되는 일을 수행하게 하는 것

- 2.1.1 단어의 의미
    - 우리의 말은 문자로 구성되며, 말의 의미는 단어로 구성. 단어를 말하자면 의미의 최소단위
    - 때문에 자연어를 컴퓨터에게 이해시키는 데는 무엇보다 단어의 의미를 이해시키는 것이 중요
    - 컴퓨터에게 단어의 의미를 이해 시키기, 단어의 의미를 잘 파악하는 표현 방법은 다음의 3가지 기법으로 살펴봄
        - 시소러스를 활용한 기법
        - 통계 기반 기법
        - 추론 기반 기법(word2vec)
    - 시소리스(유의어 사전)를 이용한 방법부터 진행

- 2.2 시소러스
    - 단어의 의미를 나타내는 방법으로는 먼저 사람이 직접 단어의 의미를 정의하는 방식을 생각할 수 있음, 그중 한 방법으로는 표준국어대사전 처럼 각각의 단어에 그 의미를 설명해 너흔ㄴ 것
    - 표준국어대사전에서 자동차 라는 단어를 찾으면 원동기를 장치하여 그 동력으로 바퀴를 굴려서 철길이나 가설된 선에 의하지 아니하고 땅위를 움직이도록 만든차 라는 설명이 나옴
    - 이런식으로 단어들을 정의해두면 컴퓨터도 단어의 의미를 이해할 수도 있음
    - 시소러스란 기본적으로는 유의어 사전으로 뜻이 같은 단어(동의어)나 뜻이 비슷한 단어(유의어)가 한 그룹으로 분류되어 있음
    - 동의어의 예
        - car = auto, automobile, machine, motorcar
    - 자연어 처리에 이용되는 시소러스에서는 단어 사이의 상위와 하위 혹은 전체와 부분들 더 세세한 관계까지 정의해둔 경우도 있음, 아래 그림처럼 각 단어의 관계를 그래프 구조로 정의

```python
# text 에 문자열을 지정하고 해당 문자열을 단어 단위로 분할
text = "You say goodbye and I say hello."
text = text.lower()
text = text.replace(".", " .")
text
```

- 위 처럼 가공하여 원래의 문장을 단어 목록 형태로 이용할 수 있게 됨, 단어 단위로 분할되어 다루기가 쉬워졌지만 단어를 텍스트 그대로 조작하기는 분편

- 위 코드에서는 파이썬의 내포(comperhension)표기를 사용하여 단어 목록에서 단어ID 목록으로 변환한 다음, 넘파이 배열로 변환

```python
def preprocess(text):
    text = text.lower()
    text = text.replace(".", " .")
    words = text.split(" ")
    word_to_id = {}
    id_to_word = {}

    for word in words:
        if word not in word_to_id:
            new_id = len(word_to_id)
            word_to_id[word] = new_id
            id_to_word[new_id] = word

    corpus = np.array([word_to_id[w] for w in words])

    return corpus, word_to_id, id_to_word
```

```python
text = "You say goodbye and I say hello."
corpus, word_to_id, id_to_word = preprocess(text)
```

- preprocess() 함수를 만들어 말뭉치의 기본적인 전처리 기능 처리
- corpus에는 단어 ID 목록, word_to_id 는 단어에서 단어id로의 딕셔너리, id_to_word는 단어 ID에서 단어로의 딕셔너리

- 이와 같이 동시발생 행렬을 활용하면 단어를 백터로 나타낼 수 있음

```python
def create_co_matrix(corpus, vocab_size, window_size=1):
    corpus_size = len(corpus)
    co_matrix = np.zeros((vocab_size, vocab_size), dtype=np.int32)

    for idx, word_id in enumerate(corpus):
        for i in range(1, window_size + 1):
            left_idx = idx - i
            right_idx = idx + i

            if left_idx >= 0:
                left_word_id = corpus[left_idx]
                co_matrix[word_id, left_word_id] += 1
            
            if right_idx < corpus_size:
                right_word_id = corpus[right_idx]
                co_matrix[word_id, right_word_id] += 1

    return co_matrix
```

- 이 함수는 먼저 co_matrix를 0으로 채워지 2차원 배열로 초기화
- 그 다음은 말뭉치의 모든 단어 각각에 대하여 윈도우에 포함된 주변 단어를 카운팅
- 이때 말뭉치의 왼쪽 끝과 오른쪽 끝 경계를 벗어나지 않는 지 확인
- 이 함수는 말뭉치가 아무리 커지더라도 자동으로 동시발생 행렬 생성

- 2.3.5 벡터 간 유사도
    - 벡터 사이의 유사도를 측정하는 방법, 대표적으로는 벡터의 내적이나 유클리드 거리 등이 있음
    - 단어 벡터의 유사도의 경우 코사인 유사도를 자주 사용, 두 벡터 x = (x_1, x_2,...,x_n), y = (y_1, y_2,...y_n) 이 있다면 코사인 유사도는 아래와 같음
    - 
    - 위 식에서 분자에는 벡터의 내적이, 분모에는 각 벡토의 노름(norm)이 등장, 노름은 벡터의 크기, 여기선,ㄴ L2 노름으로 계산, L2 노름은 벡터의 각 원소를 제곱해 더한 후 다시 제곱근을 구해 계산
    - 위 시기의 핵심은 벡터를 정규화하고 내적을 구하는 것
    - 코사인 유사도를 직관적으로 풀어보자면 두 벡터가 가리키는 방향이 얼마나 비슷한가 인데, 두 벡터의 방향이 완전히 같다면 코사인 유사도가 1이 되며, 완전히 반대라면 -1이 됨 

```python
def cos_similarity(x, y):
    nx = x / np.sqrt(np.sum(x ** 2))    # x의 정규화
    ny = y / np.sqrt(np.sum(y ** 2))    # y의 정규화

    return np.dot(nx, ny)
```

- 위 코드는 인수 x와 y는 넘파이 배열이라고 가정, 먼저 벡터 x와 y를 정규화 한 후 두 벡터의 내적을 구함
- 다만 인수로 제로벡터(원소가 모두 0인 벡터)가 들어오면 0으로 나누기 오류가 발생할 수 있음
- 이 문제를 해결하는 전통적인 방법은 나눌 때 분모에 작은 값을 더해 주는 것

```python
def cos_similarity(x, y, eps=1e-8):
    nx = x / (np.sqrt(np.sum(x ** 2)) + eps)    # x의 정규화
    ny = y / (np.sqrt(np.sum(y ** 2)) + eps)    # y의 정규화

    return np.dot(nx, ny)
```

- 여기에서 작은 값으로 1e-8을 사용했는데, 이정도 작은 값이면 일반적으로 부동소수점 계산 시 반올림되어 다른값에 흡수됨, 위의 구현에서는 이 값이 벡터의 크기에 흡수되기 땨문에 대부분의 경우 eps를 더한다고 해도 최종 결과에 영향을 주지 않지만 벡터의 크기가 0인 경우 이 값은 그대로 유지되어 0으로 나누기 오류를 회피 가능

- 2.3.6 유사 단어의 랭킹 표시
    - 어떤 단어가 검색어로 주어지면 그 검색어와 비슷한 단어를 유사도 운으로 출력하는 함수 most_cimikar()
    - most_similar(query, word_to_id, id_to_word, word_matrix, top=5)
    - 매개변수 설명
        - query: 검색어(단어)
        - word_to_id: 단어에서 단어ID로의 딕셔너리
        - id_to_word: 단어ID에서 단어로의 딕셔너리
        - word_matrix: 단어 벡터들을 한데 모은 행렬, 각 행에는 대응하는 단어의 벡터가 저장되어 있다고 가정
        - top: 상위 몇 개까지 출력할 지 지정

```python
def most_similar(query, word_to_id, id_to_word, word_matrix, top=5):
    # 1. 검색어를 꺼냄
    if query not in word_to_id:
        print("%s(을)를 찾을 수 없습니다." % query)
        return
    
    print("\n[query] " + query)
    query_id = word_to_id[query]
    query_vec = word_matrix[query_id]

    # 2. 코사인 유사도 계산
    vocab_size = len(id_to_word)
    similarity = np.zeros(vocab_size)

    for i in range(vocab_size):
        similarity[i] = cos_similarity(word_matrix[i], query_vec)

        # 3. 코사인 유사도를 기준으로 내림차순 출력
        count = 0
        for i in (-1 * similarity).argsort:
            if id_to_word[i] == query:
                continue

            print(" %s: %s" % (id_to_word[i], similarity[i]))

            count += 1
            if count >= top:
                return
```

- 위 코드는 아래 순서로 동작
    - 검색어의 단어 벡터를 꺼낸다.
    - 검색어의 단어 벡터와 다른 모든 단어 벡터와의 코사인 유사도를 각각 구한다.
    - 계산한 코사인 유사도 결과를 기준으로 값이 높은 순서대로 출력한다.
        - similarity 배열에 담긴 원소의 인덱스를 내림차순으로 정렬한 후 원소들을 출력, 이때 배열 인덱스의 정렬을 바꾸는데 사용한 argsort() 메서드는 넘파이 배열의 원소를 오름차순으로 정렬(단, 반환값은 배열의 인덱스)

```python
x = np.array([100, -20, 2])
x.argsort()
```

- 위 코드는 [100, -20, 2] 라는 넘파이 배열의 원소들을 오름차순으로 정렬, 이때 반환된 배열에 담긴 원소들은 원래 배열의 인덱스에 해당
- 즉 앞의 결과는 인덱스가 1인 원소(-20), 2인 원소(2), 0인 원소(100) 순으로 정렬
- 단어의 유사도가 큰 순서로 정렬하는 것이 목적 이었기 때문에 넘파이 배열의 각 숸소에 마이너스를 곱한 수 argsort() 메서드를 호출 하면 됨

- 2.4.1 상호정보량
    - 앞에서 작성한 동시발행 행렬의 원소는 두 단어가 동시에 발생한 횟수를 나타냄, 그러나 이 발생횟수 라는 것은 그리 좋은 특징이 아님, 고빈도 단어(많이 출현하는 단어)로 눈을 돌려보면 이유 확인 가능
    - 예를 들어 말뭉치에서 the와 car의 동시 발생은 꽤 빈번한 the car 라는 문구로 자주 발견, 한편 car와 drive는 확실히 관련이 깊은
    - 하지만 단순히 등장횟수만 보면 car는 drive보다 the와 더 관련성이 높을 수 있음
    - 이 문제를 해결하기 위해 점별 상호정보량(PMI)이라는 척도를 사용, PMI는 확률 변수 x와 y에 대해 아래 식으로 정의
    - 
    - 위 식에서 P(x)는 x가 일어날 확률, P(y)는 y가 일어날 확률, P(x, y)는 x와 y가 동시에 일어날 확률을 뜻함, PMI값이 높을수록 관련석이 높다는 의미
    - 위식을 자연어에 적용하면 P(x)는 단어 x가 말뭉치에 등장할 확률을 가리킴, 10,000개의 단어로 이루어진 말뭉치에서 the가 100번 등장한다면 P("the") = 100 / 10000 = 0.01 이 됨
    - P(x, y)는 단어 x와 y가 동시발생할 확률이므로 마찬가리조 the와 car 가 10번 동시발생 했다면 P("ths", "car") = 10 / 10000 = 0.001이 됨
    - 그럼 동시 발생 행렬(각 원소는 동시발생한 단어의 횟수)을 사용하여 위 식을 다시 써보면
    - C는 동시발생 행렬, C(x, y)는 단어 x와 y가 동시발생한 횟수, C(x)와 C(y)는 각각 단어 x와 y의 등장 횟수, 이때 말뭉치에 포함된 단어의 수를 N이라고 하면 아래 식처럼 작성 가능
    - 
    - 위 식에 따라 동시발생 행렬로 부터 PMI를 구할 수 있음, 구체적으로 계산해 보면 말뭉치의 단어 수(N)을 10,000이라 하고, the와 car와 drive가 각각 1,000번, 20번, 10번 등장했다고 가정
    - 그리고 the와 car의 동시발생 횟수 10회, car와 drive의 동시발생 횟수를 5회라 가정, 이 조건이라면 동시발생 횟수 관점에서는 car는 drive보다 the와 관련이 깊은데 PMI계산 결과는 아래와 같음
    - 
    - 
    - 이 결과에서 알수 있듯이 PMI를 이용하면 car는 the보다 drive와의 관련성이 강해짐, 이러한 결과가 나온 이유는 단어가 단독으로 출현하는 횟수가 고려되었기 때문, 여기서는 the가 자주 출현했기때문에 PMI 점수가 낮아짐
    - 여기서도 한가지 문게가 있음 PMI에서 두 단어의 동시발생 횟수가 0이라면 -무한이 된다는 점, 이 문제를 피하기 위해 실제로 구현할 때는 양의 상호정보량(PPMI)를 사용, 식은 아래와 같음
    - 
    - 위 식에 따라 PMI가 음수일때는 0으로 취급, 이제 단어 사이의 관련성을 0이상의 실수로 나타낼 수 있음

```python
def ppmi(C, verbose=False, eps=1e-8):
    M = np.zeros_like(C, dtype=np.float32)
    N = np.sum(C)
    S = np.sum(C, axis=0)
    total = C.shape[0] * C.shape[1]
    cnt = 0

    for i in range(C.shape[0]):
        for j in range(C.shape[1]):
            pmi = np.log2(C[i, j] * N / (S[j] * S[i]) + eps)
            M[i, j] = max(0, pmi)

            if verbose == True:
                cnt += 1
                if cnt % (total // 100) == 0:
                    print("%.1f%% 완료" % (100 * cnt / total))

    return M
```

- 여기에서 인수 C는 동시발생 행렬, verbose는 진행상황 출력 여부를 결정하는 플래그, 큰 말뭉치를 다룰 때 verbose를 True로 지정하면 중간중간 진행상황 표시
- 위 코드는 동시발생 행렬에 대해서는 PPMI 행렬을 구할 수 있도록 단순화해 구현
- 그리고 np.log2(0)이 음의 무한대(-inf)가 되는 사태를 막기 위해 eps 라는 작은 값을 사용

- 위 코드로 동시발생 행렬을 PPMI 행렬로 변환, 이때 PPMI 행렬의 각 원소는 0이상의 실수
- PPMI 행렬은 말뭉치의 어휘수가 증가하면 단어 벡터의 차원수도 증가한다는 문제가 있음
- 또한 이 행렬의 내용을 보면 0의 값을 가지는 원소가 많이 있음, 이는 벡터의 원소 중 많은 값이 중요하지 않은 값, 중요도가 낮은 값
- 이런 벡터는 노이즈에 약하고 견고라지 못함
- 이 문제를 대처하고자 자주 수행하는 기법이 바로 벡터의 차원 감소

    - 차원 감소는 문자 그대로 벡터의 차원을 줄이는 방법, 중요한 정보를 최대한 유지하면서 줄이는 것이 핵심, 아래 그림 처럼 데이터의 분포를 고려해 중요한 축을 찾는 일을 수행
    - 

- 위 그림의 왼쪽은 데이터들을 2차원 좌표에 표시한 모습, 오른쪽은 새로운 축을 도입하여 같은 데이터를 좌표축 하나만 표시(새로운 축을 찾을 때는 데이터가 넓게 분포되도록 고려해야 함)
- 이때 각 데이터점의 값은 새로운 축으로 사영된 값으로 변함
- 여기서 중요한 것은 가장 적합한 축을 찾아내는 일로, 1차원 값만으로도 데이터의 본질적인 차이를 구별할 수 있어야 함, 이와 같은 작업은 다차원 데이터에 대해서도 수행 가능
- 원소 대부분이 0인 행렬 또는 벡터를 희소행렬, 희소벡터라 함, 차원 감소의 핵심은 희소벡터에서 중요한 축을 찾아내어 더 적은 차원으로 다시 표현하는 것
- 차원 감소의 경롸로 원래의 희소벡터는 원소 대부분이 0이 아닌 값으로 구성된 밀집벡터가 됨, 이 조밀한 벡터가 단어의 분산 표현
- 차원을 감소시키는 방법은 여러가지가 있음, 일단 특이값분해(SVD)를 이용

- 
- 위 식과 같이 SVD는 임의의 행렬 X를 U, S, V라는 세 행렬의 곱으로 분해, 여기서 U와 V는 직교행렬이고 그 열벡터는 서로 직교, 또한 S는 대각행렬(대각성분 외에는 모두 0인 행렬), 아래 그림 참조
-

- ptb.load_data()는 데이터를 읽어 들임, 이때 인수로는 "train", "test", "valid" 중 하나 지정가능, 차례대로 훈련, 테스트, 검증용

```python
import sys
sys.path.append("..")
import numpy as np
from common.util import most_similar, create_co_matrix, ppmi
from dataset import ptb
from sklearn.utils.extmath import randomized_svd

window_size = 2
wordvec_size = 100

corpus, word_to_id, id_to_word = ptb.load_data("train")
vocab_size = len(word_to_id)
print("동시발생 수 계산 ...")

C = create_co_matrix(corpus, vocab_size, window_size)
print("PPMI 계산...")
W = ppmi(C, verbose=True)

print("SVD 계산...")

try:
    # truncated SVD (빠름)
    print("truncated SVD ")
    
    U, S, V = randomized_svd(W, n_components=wordvec_size, n_iter=5, random_state=None)
except ImportError:
    # SVD (느림)
    print("SVD ")
    U, S, V = np.linalg.svd(W)

word_vecs = U[:, :wordvec_size]
querys = ["you", "year", "car", "toyota"]

print("most_similar ")
for query in querys:
    most_similar(query, word_to_id, id_to_word, word_vecs, top=5)
```

- SVD를 수행하는 데 sklearn의 randomized_svd() 메서드를 사용, 이 메서드는 무작위 수를 사용한 Truncated SVD로, 특이값이 큰 것들만 계산하여 기본적인 SVD보다 훨씬 빠름(둘 다 테스트 해본 결과 Truncated SVD는 2분 30초, SVD는 5분이 지나도 결과가 출력되지 않음)
- 결과를 보면 우선 you라는 검색어에서는 인칭대명사 i와 we가 상위, 영어 문장에서 관용적으로 자주 같이 나오는 단어들, year의 연관어로는 month, quarter, car의 연관어로는 auto, vehicle등이 나옴
- 그리고 toyota와 연관된 단어로는 nissan, honda, lexus 등 자동자 제조업체나 브랜드가 나옴, 이처럼 단어의 의미 혹은 문법적인 관점에서 비슷한 단어들이 가까운 벡터노 나타남, 이는 인간의 직관과 비슷한 결과

- 2.5 정리
    - 자연어를 대상으로 특히 컴퓨터에게 단어의 의미를 이해시키기를 진행, 시소러스를 이요한 기법과 이어서 통계 기반 기법 진행
    - 시소러스 기반 기법에서는 단어들의 관련성을 사람이 수작업으로 하나씩 정의, 이 작업은 비효율적이고 표현력에도 한계가 있음
    - 통계 기반 기법은 말뭉치로부터 단어의 의미를 자동으로 추축, 그 의미를 벡터로 표현, 구체적으로는 단어의 동시발생 행렬을 만들고, PPMI 행렬로 변환한다음, 안정성을 높이기 위해 SVD를 이용해 차원을 감소시켜 각 단어의 분산 표현을 만들어냄, 그리고 그 분산 표현에 따르면 의미와 문법벅인 용법면에서 비슷한 단어들이 벡터공간에서도 서로 가까이 모여 있음을 확인

    - WordNet 등의 시소러스를 이용하면 유의어를 얻거나 단어 사이의 유사도를 측정하는 등 유용한 작업을 할 수 있음
    - 시소러스 기반 기법은 시소러스를 작성하는 데 엄청난 인적자원이 든다거나 새로운 단어를 대응하기 어렵다는 문제가 있음

    - 통계 기반 기법은 말뭉치 안의 각 단어에 대해서 그 단어의 주변 단어의 빈도를 집계함(동시발생 행렬)
    - 동시발생 행렬을 PPMI 행렬로 변환하고 다시 차원을 감소시킴으로써, 거대한 희소벡터를 작은 밀집벡터로 변환
    - 단어의 벡터 공간에서는 의미가 가까운 단어는 그 거리도 가까울 것으로 기대됨
